```python
'''
MOVE TO SCRIPT!!!
'''


def load_fao_gaul_data(country_name="Zambia", admin_level="level0"):
    """
    Load FAO GAUL dataset for a given country and administrative level using Google Earth Engine.
    
    Parameters:
    - country_name (str): Name of the country (default: "Zambia").
    - admin_level (str): Administrative level to load ("level0", "level1", "level2", etc.).
    
    Returns:
    - gpd.GeoDataFrame: A GeoDataFrame containing the administrative boundaries.
    - list: A list of available administrative regions (ADM names).
    """
    try:
        # Initialize Earth Engine if not already initialized
        ee.Initialize()

        # Load FAO GAUL dataset for the specified admin level
        roi = ee.FeatureCollection(f"FAO/GAUL/2015/{admin_level}")
        roi = roi.filter(ee.Filter.eq('ADM0_NAME', country_name))

        # Determine the correct attribute for region names
        admin_name_key = "ADM2_NAME" if admin_level == "level2" else "ADM1_NAME" if admin_level == "level1" else "ADM0_NAME"

        # Get available administrative regions
        counties_list = roi.aggregate_array(admin_name_key).getInfo()

        # Convert to GeoDataFrame
        gdf = geemap.ee_to_gdf(roi)

        return gdf, counties_list

    except Exception as e:
        print(f"Error loading FAO GAUL data: {e}")
        return None, None
```

```python
# Load Zambia at Level 2
zambia_gdf_2, counties_list_level2 = load_fao_gaul_data(admin_level="level2")
print(counties_list_level2)
```
